## 1.1 创建数组

NumPy 提供了多种创建数组的方法：
- `np.zeros(shape)` — 创建全零数组
- `np.full(shape, value)` — 创建指定值的数组
- `np.array(list)` — 从 Python 列表创建

```python
import numpy as np

# 全零数组
a = np.zeros((2, 3))
print("全零数组:")
print(a)

# 全一数组
b = np.full((2, 3), 1)
print("\n全一数组:")
print(b)
```

从 Python 列表创建数组。与普通列表相比，NumPy 数组支持高效的矩阵运算，是神经网络计算的基础。

```python
list1 = [[1, 2],
         [3, 4],
         [5, 6]]
array1 = np.array(list1)
print("从列表创建的数组:")
print(array1)
print("形状:", array1.shape)  # (3, 2) — 3行2列
```

## 1.2 索引与切片

NumPy 数组使用 `[行, 列]` 的方式索引，支持切片操作：
- `array[i, j]` — 第 i 行第 j 列的元素
- `array[:, j]` — 第 j 列（所有行）
- `array[i, :]` — 第 i 行（所有列）

```python
# 取单个元素：第1行第0列（索引从0开始）
print("第1行第0列:", array1[1, 0])  # 输出: 3

# 取一列：所有行的第0列
print("第0列:", array1[:, 0])  # 输出: [1 3 5]

# 取一行：第2行所有列
print("第2行:", array1[2, :])  # 输出: [5 6]
```

## 1.3 矩阵运算

神经网络的核心计算就是矩阵乘法。NumPy 提供了丰富的线性代数运算。

```python
matrix1 = np.array([[1, 2],
                    [3, 4]])
matrix2 = np.array([[5, 6],
                    [7, 8]])
```

### 矩阵点乘（矩阵乘法）

这是神经网络前向传播的核心运算：$C = A \cdot B$

注意：矩阵点乘**不是**逐元素相乘，而是行列对应相乘再求和。

```python
# 矩阵乘法：np.dot(A, B) 或 A @ B
print("矩阵点乘 (np.dot):")
print(np.dot(matrix1, matrix2))

# Hadamard 乘积（逐元素相乘）：A * B
print("\n逐元素相乘 (Hadamard):")
print(matrix1 * matrix2)
```

### 转置

转置将矩阵的行变成列、列变成行：$(A^T)_{ij} = A_{ji}$

```python
print("原矩阵:")
print(matrix1)
print("\n转置:")
print(matrix1.T)
```

### 行列式与逆矩阵

- **行列式**：衡量矩阵对空间的缩放因子
- **逆矩阵**：$A \cdot A^{-1} = I$，类似于数字的倒数

```python
print("行列式:", np.linalg.det(matrix1))
print("\n逆矩阵:")
print(np.linalg.inv(matrix1))
```

## 1.4 聚合函数与 axis 参数

`axis` 参数决定了聚合的方向。记住这个规则：

| 操作 | axis 值 | 含义 |
| --- | --- | --- |
| `np.mean(m, axis=0)` | 0 | 沿行方向压缩 → **每列的平均** |
| `np.mean(m, axis=1)` | 1 | 沿列方向压缩 → **每行的平均** |
| `np.mean(m)` | 无 | 所有元素的平均 |

```python
matrix = np.array([[1, 2],
                   [3, 4]])

print("所有元素平均:", np.mean(matrix))       # 2.5
print("每列平均 (axis=0):", np.mean(matrix, axis=0))  # [2. 3.]
print("每行平均 (axis=1):", np.mean(matrix, axis=1))  # [1.5 3.5]
```

## 1.5 np.maximum vs np.max

这两个函数名字相似但用途完全不同：

- `np.max(array)` — 返回数组中的**最大值**（标量）
- `np.maximum(a, b)` — 逐元素取 a 和 b 中的**较大值**（数组）

`np.maximum(0, x)` 就是 **ReLU 激活函数**，在后续课程中会频繁使用。

```python
matrix = np.array([[1, 2],
                   [3, 4]])

# np.max: 数组中的最大值
print("数组最大值 np.max:", np.max(matrix))  # 4

# np.maximum(0, x): 负数变0，正数不变 → ReLU!
print("\nnp.maximum(0, matrix) [ReLU]:")
print(np.maximum(0, matrix))

# np.maximum: 逐元素取两个数组的较大值
matrix2 = np.array([[5, 1],
                    [2, 8]])
print("\nnp.maximum(matrix, matrix2):")
print(np.maximum(matrix, matrix2))  # [[5, 2], [3, 8]]
```
